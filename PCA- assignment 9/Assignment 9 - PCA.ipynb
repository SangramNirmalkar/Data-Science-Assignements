# Dimentionality reduction with PCA

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X = df.drop(columns=['Type'])

scaler = StandardScaler()

x_scaled =scaler.fit_transform(X)
```

```python
x_scaled
```

```python
X_scaled_df = pd.DataFrame(x_scaled,columns=X.columns)
```

```python
X_scaled_df
```

```python
X_scaled_df.describe()
```

after standardization the mean was observed around zero and standard deviation was around 1.  this means our data is standardized properly and now ready to use for modeling.

```python
pca = PCA()
X_pca_full =pca.fit_transform(x_scaled)
explained_variance = pca.explained_variance_ratio_
explained_variance
```

```python
plt.figure(figsize=(10,6))
plt.plot(range(1,len(explained_variance)+1), explained_variance, marker='o')
plt.xlabel('Principal components')
plt.ylabel('Explained variance ration')
plt.title('Scree plot')
plt.show()
```

you can see the first few components explained a lot of variance in the data, it indicates that there is a redundancy in the data. after 4-5 components the new components adds very little information. basically our 4-5 components expalians a lot of  variance

```python
import numpy as np
import matplotlib.pyplot as plt

cumulative_variance = np.cumsum(explained_variance)

plt.figure()
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker='o')
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")
plt.axhline(y=0.8, linestyle='--') 
plt.show()
```

based on the cumulative variance plot approximately 80 percent of  total variance is captured by first 4 components. beyond this point additional components contribute very less.. therefore we selecy 4v components as optimal balance between dim reduction and info retention.

```python
pca = PCA(n_components=4)
x_pca = pca.fit_transform(x_scaled)
```

```python
x_pca_df = pd.DataFrame(x_pca,columns=['PC1','PC2','PC3','PC4'])
x_pca_df.head(10)
```

```python
pca.explained_variance_ratio_.sum()
```

as you can see the pca reduction to 4 components almost capture 73 percent of data, which is good . which explains almost 80 percent of variance in the data.

# Clustering on original data

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
inertia = []
k_range = range(2,11)

for k in k_range:
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(x_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure()
plt.plot(k_range, inertia, marker='o')
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method (Original Data)")
plt.show()
```

```python
for k in range(2,7):
    kmeans = KMeans(n_clusters=k,random_state=42)
    labels =kmeans.fit_predict(x_scaled)
    score = silhouette_score(x_scaled,labels)
    print(f'k={k}, silhouette score={score:.3f}')
```

    elbow appears at k=3  and silhouette score is also relatively high for k =3.
    the optimal numebr of clusters was determined for k-means using elbow method and silhouette score

```python
kmeans_org = KMeans(n_clusters=3, random_state=42)
cluster_label_org = kmeans_org.fit_predict(x_scaled)
```

```python
plt.figure()
plt.scatter(
    x_scaled[:, 0],
    x_scaled[:, 1],
    c=cluster_label_org
)
plt.xlabel("Feature 1 (standardized)")
plt.ylabel("Feature 2 (standardized)")
plt.title("K-means Clustering on Original Data")
plt.show()
```

```python
silhouette_org = silhouette_score(x_scaled,cluster_label_org)
silhouette_org
```

the silhouette score indicates moderate cluster separation , indicating the overlap that may exist in the original data.

 kmeans  clustering in the standardized original features resulted in moderately separated cluster, indicating the presense of underlying structure in the data.

# PCA Clustering

```python
inertia_pca = []
k_range = range(2,11)

for k in k_range:
    kmeans_pca = KMeans(n_clusters=k,random_state=42)
    kmeans_pca.fit(x_pca_df)
    inertia_pca.append(kmeans_pca.inertia_)
```

```python
plt.figure()
plt.plot(k_range, inertia_pca, marker='o')
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia_pca")
plt.title("Elbow Method (PCA Data)")
plt.show()
```

```python
for k in range(2,7):
    kmeans_pca = KMeans(n_clusters=k,random_state=42)
    labels_pca =kmeans_pca.fit_predict(x_pca_df)
    score_pca = silhouette_score(x_pca_df,labels_pca)
    print(f'k={k}, silhouette score={score_pca:.3f}')
```

```python
kmeans_pca_final = KMeans(n_clusters=3,random_state=42)
cluster_labels_pca = kmeans_pca_final.fit_predict(x_pca_df)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(
    x_pca_df['PC1'],
    x_pca_df['PC2'],
    c=cluster_labels_pca
)
plt.xlabel('principal Component 1')
plt.ylabel('principal component 2')
plt.title('K means clustering on pca data')
plt.show()
```

```python
silhouette_pca = silhouette_score(x_pca_df,cluster_labels_pca)
silhouette_pca
```

compared to the clustering on the original standardized features, kmeans clustering on pca transformed data resulted in a higher silhouette score, indicating improved cluster separation after pca dimentionality reduction.

# comparison

```python
comparison_df = pd.DataFrame({
    'Dataset':['Original_scaled_data','PCA_scaled/Transformed_data'],
    'Number_of_features':[x_scaled.shape[1],x_pca_df.shape[1]],
    'silhouette_score':[silhouette_org,silhouette_pca]
})

comparison_df
```

you can see after pca dimentionality reduction the silhouette score improved a lot compared to the 13 dimentional original data. clusters are more better separated after pca.

dimentionality reduction on pca data had a positive impact on clustering performance. by reducing the dataset from  13 original features to 4 principal components while retaining most of the variance, pca minimized variance and most of the redundancy.

The tradeoffs are while performing clustering on the original data retains full features interpretability, no information compression but it is sensitive ti correlated features, computationally heavier. While in pca reduced dimentionailty and noise , improved clustering performance and faster computations are benefits along with  better visualizations.

# Conclusion

the assignement demonstrated the k means clustering performance is influenced not only by the algorithm but also by data represnetation. while clusetring on the original standardized dataset revealed some structure, applying pca before clustering significantly imprived cluster searation . dimentionality reduction effecively reduced noise and redundancy while preserving most of the important information.

In practical  data analysis scenarios, dataset often contains high dimentional and correlated features. applying pca before clustering can enhance performance, reduce computational tasks, and improve visualizations. this is particularly usefull in exploratory data analysis.

recommendations to use pca: when dataset is high dimentional , features are strongly correlated and improevd clustring performance and visualization are high priority then using pca is a must.
