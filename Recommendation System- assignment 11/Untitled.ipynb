# recommendation system

```python
def recommendation_anime(title, top_n=5):
    if title not in  indices:
        return 'Anime not found in dataset.'
    idx =indices[title]
    sim_score = list(enumerate(cosine_sim[idx]))
    sim_score = sorted(sim_score,key= lambda x: x[1], reverse=True)
    sim_score = sim_score[1:top_n+1]
    anime_indices = [i[0] for i in sim_score]
    return df[['name','genre','rating']].iloc[anime_indices]
```

```python
recommendation_anime('Bleach')
```

```python
title ='Naruto'
idx = indices[title]

scores = list(enumerate(cosine_sim[idx]))

scores = sorted(scores, key =lambda x: x[1], reverse=True)

scores[:10]
```

```python
scores = scores[1:11] # removing the self threshold for anime's
scores
```

```python
for i , score in scores:
    print(df.iloc[i]['name'], round(score,3))
```

```python
def recommend_anime_threshold(title,threshold = 0.5):
    if title not in indices:
        return "Anime not found"
    idx = indices[title]
    score = scores[1:]
    filtered = [(i,s) for i , s in score if s> threshold]
    filtered = sorted(filtered, key=lambda x: x[1], reverse=True)
    anime_indices = [i[0] for i in filtered]
    
    return df[['name', 'genre','rating']].iloc[anime_indices]
```

```python
recommend_anime_threshold('Naruto', threshold= 0.75)
```

```python
recommendation_anime('Death Note')
```

since this is an unsupervised learning , content based recommendation system,labels are not available 

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42
)
```

```python
train_indices = pd.Series(
    train_df.index,
    index=train_df['name']
).drop_duplicates()
```

```python
def is_relevant(test_genre, recommended_genre):
    test_set = set(test_genre.split(', '))
    rec_set = set(recommended_genre.split(', '))
    return len(test_set.intersection(rec_set)) > 0
```

```python
def evaluate_single_anime(test_row, top_n=5):
    title = test_row['name']
    
    if title not in indices:
        return None
    
    recommendations = recommendation_anime(title, top_n)
    
    if isinstance(recommendations, str):
        return None
    
    relevant_count = 0
    
    for _, row in recommendations.iterrows():
        if is_relevant(test_row['genre'], row['genre']):
            relevant_count += 1
    
    precision = relevant_count / top_n
    
    # Recall approximation:
    # number of relevant recommendations / total possible relevant anime
    total_relevant = 0
    for _, row in train_df.iterrows():
        if is_relevant(test_row['genre'], row['genre']):
            total_relevant += 1
    
    recall = relevant_count / total_relevant if total_relevant > 0 else 0
    
    if precision + recall == 0:
        f1 = 0
    else:
        f1 = 2 * precision * recall / (precision + recall)
    
    return precision, recall, f1
```

```python
sample_test = test_df.sample(100, random_state=42)

precisions = []
recalls = []
f1_scores = []

for _, row in sample_test.iterrows():
    result = evaluate_single_anime(row, top_n=5)
    
    if result:
        p, r, f = result
        precisions.append(p)
        recalls.append(r)
        f1_scores.append(f)
```

```python
avg_precision = sum(precisions) / len(precisions)
avg_recall = sum(recalls) / len(recalls)
avg_f1 = sum(f1_scores) / len(f1_scores)

avg_precision, avg_recall, avg_f1
```

# conclusion

as you can see the precision score indicates that nearly half of the recommended anime share genre similarity with the target anime. however recall is very low because the total number of genre matching anime in the dataset is large, while only a small number of recommendations top 5 are generated . this is a known limitation when applying traditional reacall metrics to the recommnender system. this recommender system is unsupervised. Since explicit user relevance labels were unavailable, genre overlap was used as a proxy for relevance

# interview questions

1. Can you explain the difference between user-based and item-based collaborative filtering?

    user-based collaborative filtering:
    
        finds user with similar patterns, uses similarity measures(cosine similarity) and recommends items liked by similar users. for example: user A and user B both liked Naruto , one piece. and user B liked Bleach so recommending Bleach to user A might also be good since probablity of A liking bleach is high.
        
     item-based collaborative filtering:
         
         in item based collaborative filtering it finds the items with similar user interaction patterns and recommends items similar to what the users already liked. for example : if naruto and one piece are liked by the same users . while users liked Naruto then we can recommend one piece to the user as well.

2. What is collaborative filtering, and how does it work?
    
    Collaborative filtering is recommendation technique that makes susggestion based on user behaviour rather than item content.It works by finding patterns in user interactions, such as ratings, likes, or clicks, and assumes that users who behaved similarly in the past will prefer similar items in the future.
    
    there are two main types:
       
       1. User-based, which recommends items liked by similar users
       
       2. item-based, which recommends item  similar to what the user had already liked.
       
       
     How it works:
       
       1.Collect user–item interaction data (ratings, views, purchases)
       
       2.Measure similarity between users or items
       
       3.Identify the most similar users/items
       
       4.Recommend items the user hasn’t interacted with yet
