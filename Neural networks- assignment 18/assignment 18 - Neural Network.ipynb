```python
X = df.drop('letter', axis=1)
y = df['letter']
```

```python
# preprocessing
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

# model implementation

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled,y,
                                                   test_size=0.2,
                                                   random_state=42,
                                                   stratify=y)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_test_enc = le.transform(y_test)
```

we applied label encoding to convert the alphabet labels into numerical form suitable for neural network training

# Model Building- base

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python
model = Sequential()

# Input layer
model.add(Dense(128,activation='relu', input_shape=(X_train.shape[1],)))
# hidden
model.add(Dense(64,activation='relu',))
#output
model.add(Dense(64,activation='softmax'))
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])
```

```python
history = model.fit(
            X_train,
            y_train_enc,
            epochs = 20,
            batch_size=32,
            validation_split=0.1)
```

```python
test_loss,test_acc = model.evaluate(X_test,y_test_enc)

print('Test accuracy:',test_acc)
print('Test Loss:', test_loss)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred_probs = model.predict(X_test)
y_pred =np.argmax(y_pred_probs,axis=1)

print(confusion_matrix(y_test_enc,y_pred))
print(classification_report(y_test_enc,y_pred))
```

from above results we can see that class 7 have the lowest recall. the model struggles that to recognize the letter H or class 7 in this dataset.

The neural network achieved an overall accuracy of 94% on the test dataset, indicating strong performance for a 26-class classification problem.
Most classes show precision and recall above 90%, demonstrating consistent classification ability across alphabets.
Some classes exhibit slightly lower recall (e.g. class 7), suggesting the model occasionally confuses visually similar letters.
Overall, the confusion matrix shows strong diagonal dominance, confirming that the model correctly classifies the majority of samples.

# Hyperparameter tuning

```python
model2 = Sequential()

model2.add(Dense(256, activation='relu', input_shape=( X_train.shape[1],)))
model2.add(Dropout(0.3))
model2.add(Dense(128, activation='relu'))
model2.add(Dropout(0.3))
model2.add(Dense(26, activation='softmax'))
```

```python
model2.compile(
            optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy'])
```

```python
history2 = model2.fit(
            X_train,
            y_train_enc,
            epochs=30,
            batch_size=32,
            validation_split=0.1)
```

# Evaluation- final model

```python
test_loss2, test_acc2 = model2.evaluate(X_test, y_test_enc)
print("Improved Test Accuracy:", test_acc2)
```

```python
y_pred_probs2 = model2.predict(X_test)
y_pred2 = np.argmax(y_pred_probs2, axis=1)
cm2 = confusion_matrix(y_test_enc, y_pred2)
print(cm2)
```

```python
print(classification_report(y_test_enc, y_pred2))
```

To improve the neural network performance, the architecture was modified by increasing the number of neurons and adding dropout layers to reduce overfitting. The number of training epochs was also increased.

The tuned model achieved a test accuracy of 96.4%, compared to 94.5% from the baseline model.

The addition of dropout reduced overfitting and improved the model’s ability to generalize to unseen data. This demonstrates the importance of hyperparameter tuning in neural network optimization.

The tuned neural network achieved an overall accuracy of 96% on the test dataset. Precision, recall, and F1-score values are consistently high across all 26 classes, indicating balanced model performance.
Compared to the baseline model, the tuned model significantly improved recall for previously difficult classes, demonstrating better generalization and reduced overfitting.
The results confirm that hyperparameter tuning and dropout regularization improved the robustness of the neural network.
